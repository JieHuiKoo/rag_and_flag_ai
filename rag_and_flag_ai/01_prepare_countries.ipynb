```python
import pandas as pd
import wikipedia
```

```python
# read the list of countries
with open("rag_and_flag_ai/list_of_countries.txt", 'r') as file:
    countries_str = file.read().split(',')
    print(countries_str)
```

```python
# get list of country names, excluding empty string
countries = [ country.split("|")[1] for country in countries_str if country != "" and "|" in country ]
print(countries)
```

```python
# download country articles from Wikipedia
countries_subset = countries[:10]
df = pd.DataFrame({
    "country": countries_subset,
    "text": [""] * len(countries_subset)
})
print(df)
```

```python
for idx, row in df.iterrows():
    country = row['country']
    # update country text
    # df.at[idx, 'text'] = wikipedia.search(country)[0]
    print(country)
    df.at[idx, 'text'] = wikipedia.page(title=country,auto_suggest=False).content


print(df)
```

```python
# Export to CSV
df.to_csv("rag_and_flag_ai/countries_subset.csv", index=False)
```
